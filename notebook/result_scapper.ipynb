```python
import wandb
import pandas as pd
import os
from tqdm.notebook import tqdm
```

```python
api = wandb.Api(timeout=600)
```

```python
project_list = [
                "eval_multi_cont_raw_pcgrl",
                "eval_single_cont_raw_pcgrl",]
diversity=True
```

```python
def get_process_config(config, run_id):
    config_dict = deepcopy(config)
    config_dict['run_id'] = run_id
    # remove keys
    
    # flatten the keys
    flatten_keys = ['encoder']
    
    for key in flatten_keys:
        sub_dict = config_dict[key]
        sub_dict = eval(sub_dict)
        for sub_key, value in sub_dict.items():
            config_dict[f'{key}.{sub_key}'] = value

    remove_keys = ['lr', 'GAMMA', 'is_3d', 'n_eps', 'agents', 'n_bins', 'n_gpus', 'VF_COEF', 'encoder', 'CLIP_EPS', 'ENT_COEF', '_img_dir', '_vid_dir', 'arf_size', 'env_name', 'eval_dir', 'n_agents', 'vrf_size', 'ANNEAL_LR', 'act_shape', 'ckpt_freq', 'eval_freq', 'eval_seed', 'evaluator', 'map_width', 'num_steps', 'overwrite', 'pinpoints', 'wandb_key', 'GAE_LAMBDA', '_numpy_dir', 'activation', 'change_pct', 'initialize', 'multiagent', 'n_freezies', 'NUM_UPDATES', 'empty_start', 'hidden_dims', 'n_eval_envs', 'n_eval_maps', 'profile_fps', 'render_freq', 'ctrl_metrics', 'n_render_eps', 'reward_every', 'wandb_entity', 'wandb_resume', 'MAX_GRAD_NORM', 'eval_map_path', 'update_epochs', 'wandb_project', 'MINIBATCH_SIZE', 'eval_map_types', 'eval_map_width', 'NUM_MINIBATCHES', 'max_board_scans', 'metrics_to_keep', 'static_tile_prob', 'current_iteration', 'gif_frame_duration', 'randomize_map_shape', 'timestep_chunk_size', 'eval_max_board_scans', 'DEBUG', 'eval_randomize_map_shape']
    
    for key in remove_keys:
        config_dict.pop(key, None)
    # change the values to str
    for key, value in config_dict.items():
        config_dict[key] = str(value)
        
    return config_dict
```

```python
import json
from copy import deepcopy

# mkdir
os.makedirs('results', exist_ok=True)

for project_name in project_list:
    runs = api.runs(f"your_wandb_entity/{project_name}")

    for run in tqdm(runs, desc=project_name):

        # if the run id directory exists, skip
        if os.path.exists(f"results/{project_name}/{run.id}"):
            print(f"results/{project_name}/{run.id} exists")
            continue

        try:
            run_name = run.id
            config_dict = get_process_config(run.config, run_name)

            _run = wandb.init()

            artifact = _run.use_artifact(
                f'your_wandb_entity/{project_name}/run-{run_name}-raw:v0')
            artifact.download(root=".temp")

            # make the directory with project name
            os.makedirs(f"results/{project_name}", exist_ok=True)
            os.makedirs(f"results/{project_name}/{run_name}", exist_ok=True)
            # move the file to the directory
#            os.rename(".temp/diversity.table.json", f"results/{project_name}/{run_name}/diversity.json")
            os.rename(".temp/raw.table.json", f"results/{project_name}/{run_name}/raw.json")

            # load to dataframe and save to csv

            data_dict = json.load(open(f"results/{project_name}/{run_name}/raw.json"))

            raw_df = pd.DataFrame(data_dict["data"], columns=data_dict["columns"])

            config_df = pd.DataFrame([{f"config.{k}": v for k, v in config_dict.items()}])
            config_df = pd.concat([config_df] * len(raw_df), ignore_index=True)

            raw_df = pd.concat([raw_df.reset_index(drop=True), config_df], axis=1)

            raw_df.to_csv(f"results/{project_name}/{run_name}/raw.csv", index=False)

            if diversity:
                data_dict = json.load(open(f"results/{project_name}/{run_name}/diversity.json"))

                div_df = pd.DataFrame(data_dict["data"], columns=data_dict["columns"])

                div_df = pd.concat([div_df.reset_index(drop=True), config_df], axis=1)
                div_df.to_csv(f"results/{project_name}/{run_name}/diversity.csv", index=False)

            # save config
            with open(f"results/{project_name}/{run_name}/config.json", 'w') as f:
                json.dump(run.config, f)
        except Exception as e:
            print(e)
```
