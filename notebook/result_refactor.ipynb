```python

import os
import shutil
import re
from tqdm.notebook import tqdm
```

```python
#config

RESULT_DIR = 'results'

BASE_DIR_LIST = [
    'eval_single_cont_raw_pcgrl',
    'eval_multi_cont_raw_pcgrl',
]
TARGET_DIR = 'exp_0428'

default_encoder = 'cont_raw'

print_log =True
```

```python
for BASE_DIR in BASE_DIR_LIST:
    base_path=os.path.join(RESULT_DIR, BASE_DIR)
    target_path=os.path.join(RESULT_DIR, TARGET_DIR)
    os.makedirs(target_path, exist_ok=True)

    for run_name in tqdm(os.listdir(base_path), desc=base_path):

        folder_path = os.path.join(base_path, run_name)
        if not os.path.isdir(folder_path):
            continue  # 파일 무시

        # inst- 추출
        instruct_match = re.search(r"inst-([^\s_]+(?:_[^\s_]+)*?)(?:_(enc|tr|model|exp|norm|s)-|$)", run_name)
        inst = instruct_match.group(1) if instruct_match else "unknown"

        # exp- 추출
        def_match = re.search(r"exp-([^\s_]+(?:_[^\s_]+)*?)(?:_(enc|tr|model|exp|norm|s)-|$)", run_name)
        def_name = def_match.group(1) if def_match else "base"
        def_name=def_name[4:]#def_ 제거

        if def_name == 'cls_st_mk':
            def_name = "mk"
        if def_name == '':
            def_name = default_encoder

        dest_path = os.path.join(target_path, inst, def_name, run_name)

        try:
            shutil.move(folder_path, dest_path)
            if print_log:
                print(f"✅ Moved: {folder_path} ➝ {dest_path}")
        except Exception as e:
            if print_log:
                print(f"❌ Failed to move {folder_path}: {e}")

    #빈 base_폴더 삭제
    if not os.listdir(base_path):
        try:
            os.rmdir(base_path)
            if print_log:
                print(f"🧹 Removed empty base directory: {base_path}")
        except Exception as e:
            if print_log:
                print(f"⚠️ Failed to remove base directory: {e}")
```
